```python
from IPython.display import HTML, display, clear_output

# YOUR SECURE WORKER URL
WORKER_URL = "https://medical-portfolio-proxy.bakwasandsony.workers.dev" 

js_code = f"""
<script>
    // Create the big button
    const btn = document.createElement('button');
    btn.innerHTML = "🤖 Click Here for AI Feedback on Your Reflection";
    btn.style.backgroundColor = "#007bff";
    btn.style.color = "white";
    btn.style.padding = "15px 30px";
    btn.style.fontSize = "18px";
    btn.style.border = "none";
    btn.style.borderRadius = "8px";
    btn.style.cursor = "pointer";
    btn.style.marginTop = "10px";
    btn.style.fontWeight = "bold";
    btn.style.boxShadow = "0 4px 8px rgba(0,0,0,0.2)";

    // Create a div to show the output
    const outputDiv = document.createElement('div');
    outputDiv.style.marginTop = "20px";

    // Add them to the notebook
    const container = document.createElement('div');
    container.appendChild(btn);
    container.appendChild(outputDiv);
    document.currentScript.parentNode.appendChild(container);

    // What happens when the button is clicked
    btn.onclick = async function() {{
        const studentReflection = prompt("Please paste your 300-word reflection here:");
        
        if (studentReflection && studentReflection.trim() !== "") {{
            btn.disabled = true;
            btn.innerHTML = "🧠 AI is reviewing your reflection... Please wait...";
            outputDiv.innerHTML = "";

            try {{
                const response = await fetch("{WORKER_URL}", {{
                    method: "POST",
                    headers: {{ "Content-Type": "application/json" }},
                    body: JSON.stringify({{ reflection: studentReflection }})
                }});
                
                const data = await response.json();
                
                if (data.error) {{
                    outputDiv.innerHTML = "<b style='color:red;'>Error: " + data.error + "</b>";
                }} else if (data.choices && data.choices[0]) {{
                    const aiFeedback = data.choices[0].message.content;
                    outputDiv.innerHTML = "<h4>🤖 AI Feedback:</h4><pre style='white-space: pre-wrap; background: #f4f4f4; padding: 10px; border-radius: 5px; border-left: 5px solid #007bff;'>" + aiFeedback + "</pre>";
                }} else {{
                    outputDiv.innerHTML = "<b style='color:red;'>Unexpected response from server.</b>";
                }}
            }} catch (error) {{
                outputDiv.innerHTML = "<b style='color:red;'>Network Error: " + error + "</b>";
            }}
            
            btn.disabled = false;
            btn.innerHTML = "🤖 Click Here for AI Feedback on Your Reflection";
        }} else {{
            alert("No reflection provided. The AI will not run.");
        }}
    }};
</script>
"""

display(HTML(js_code))
```
