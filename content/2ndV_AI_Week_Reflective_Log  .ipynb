```python
from IPython.display import HTML, display

# YOUR NEW SECURE WORKER URL
WORKER_URL = "https://medical-portfolio-proxy.bakwasandsony.workers.dev" 

js_code = f"""
<script>
    const studentReflection = prompt("Paste your 300-word reflection here for AI feedback:");

    if (studentReflection && studentReflection.trim() !== "") {{
        const outputDiv = document.createElement('div');
        outputDiv.innerHTML = "<b>🧠 AI is reviewing your reflection... Please wait 3 seconds...</b>";
        document.currentScript.parentNode.appendChild(outputDiv);

        fetch("{WORKER_URL}", {{
            method: "POST",
            headers: {{ "Content-Type": "application/json" }},
            body: JSON.stringify({{ reflection: studentReflection }})
        }})
        .then(response => response.json())
        .then(data => {{
            if (data.error) {{
                outputDiv.innerHTML = "<b style='color:red;'>Error: " + data.error + "</b>";
            }} else if (data.choices && data.choices[0]) {{
                const aiFeedback = data.choices[0].message.content;
                outputDiv.innerHTML = "<h4>🤖 AI Feedback:</h4><pre style='white-space: pre-wrap; background: #f4f4f4; padding: 10px; border-radius: 5px;'>" + aiFeedback + "</pre>";
            }} else {{
                outputDiv.innerHTML = "<b style='color:red;'>Unexpected response from server.</b>";
            }}
        }})
        .catch(error => {{
            outputDiv.innerHTML = "<b style='color:red;'>Network Error: " + error + "</b>";
        }});
    }} else {{
        alert("No reflection provided. The AI will not run.");
    }}
</script>
"""

display(HTML(js_code))
```
